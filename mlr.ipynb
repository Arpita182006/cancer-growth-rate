```python
import numpy as np
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
```

```python
ds = pd.read_csv("cancer_growth.csv")
ds
```

```python
ds.info()
```

```python
ds.columns
```

```python
#splitting data into independent and dependent variables

x=ds[['Age', 'Gender', 'Previous_Tumor_Size',
       'Time_Since_Diagnosis', 'Stage', 'Genetic_Risk', 'Smoking_History',
       'Alcohol_Consumption', 'BMI',
       'Treatment_Chemotherapy', 'Treatment_Radiotherapy',
       'Treatment_Surgery']]
y=ds[['Tumor_Growth_Rate']]
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import  r2_score

print("✅ Multiple Linear Regression Results")
print("R² Score:", r2_score(y_test, y_pred))
```

```python
#user input
print("\nEnter patient details:")

age = int(input("Age: "))
gender = int(input("Gender (Male=0, Female=1): "))
previous_size = float(input("Previous Tumor Size: "))
time_since = int(input("Time Since Diagnosis: "))
stage = int(input("Stage (I=1, II=2, III=3, IV=4): "))
genetic_risk = int(input("Genetic Risk (Low=0, Medium=1, High=2): "))
smoking = int(input("Smoking History (No=0, Yes=1): "))
alcohol = int(input("Alcohol Consumption (Low=0, Medium=1, High=2): "))
bmi = float(input("BMI: "))

# Treatment one-hot encoding (only one should be 1)
treatment_chemo = int(input("Treatment Chemotherapy (Yes=1, No=0): "))
treatment_radio = int(input("Treatment Radiotherapy (Yes=1, No=0): "))
treatment_surgery = int(input("Treatment Surgery (Yes=1, No=0): "))

user_data = pd.DataFrame([[
    age, gender, previous_size, time_since, stage, genetic_risk, smoking,
    alcohol, bmi, treatment_chemo, treatment_radio, treatment_surgery
]], columns=x.columns)

prediction = model.predict(user_data)[0]


print("\n🎯 Predicted Tumor Growth Rate:",prediction)
```
